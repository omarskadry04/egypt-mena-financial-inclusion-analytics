# Egypt & MENA Digital Financial Inclusion Opportunity Analytics

## 1. Notebook purpose
Prepare a small Central Bank of Egypt (CBE) context table for Egypt. Findex remains the main consumer dataset and IMF FAS the cross-country infrastructure source. This CBE table is kept separate.

## 2. Source documents and methodology
Primary source: CBE, *Financial Inclusion and Payment Systems & Services Indicators*, December 2025, pp. 2 and 4 (PDF page numbers). Cross-check: CBE, *Progress of Financial Inclusion Indicators in Egypt*, December 2025, p. 1, for the 2025 inclusion rates. Copies are in `../data/raw/`. Values below are transcribed from printed chart labels and callouts; no chart positions were estimated and no missing years were filled. Percentages are converted to proportions.

```python
import pandas as pd
from pathlib import Path

output_path = Path("../data/processed/cbe_egypt_context.csv")
```

## 3. CBE definition and comparability
CBE financial inclusion means ownership of **active transactional accounts** by citizens aged 15 or older, including accounts at banks and Egypt Post, mobile wallets, and prepaid cards (main report, p. 2). Youth refers to ages 15–35. CBE rates must not be treated as identical to World Bank Findex account ownership: definitions and data collection methods differ. Direct numerical differences between CBE financial-inclusion rates and Findex account-ownership rates should not be interpreted as adoption gaps or changes because the measures are not methodologically equivalent. Use these observations for Egypt-specific trends and context, never as substitutes for cross-country Findex KPIs.

## 4. Overall financial inclusion trend
The printed annual rate labels are on the main report’s p. 2.

```python
overall_rates_pct = {2016: 27.4, 2017: 33.8, 2018: 38.5, 2019: 50.1, 2020: 52.7, 2021: 56.2, 2022: 64.8, 2023: 70.7, 2024: 74.8, 2025: 77.6}
overall_rates = pd.Series(overall_rates_pct, name="percent")
overall_rates.to_frame()
```

## 5. Women’s financial inclusion trend
The women’s rate labels are on the main report’s p. 2.

```python
women_rates_pct = {2016: 19.1, 2017: 24.3, 2018: 28.2, 2019: 44.4, 2020: 46.4, 2021: 50.2, 2022: 57.6, 2023: 62.7, 2024: 68.8, 2025: 71.4}
women_rates = pd.Series(women_rates_pct, name="percent")
women_rates.to_frame()
```

## 6. Youth financial inclusion trend
Youth is ages 15–35; the published series begins in 2020 (main report, p. 2).

```python
youth_rates_pct = {2020: 36.3, 2021: 39.7, 2022: 46.6, 2023: 51.5, 2024: 53.1, 2025: 56.8}
youth_rates = pd.Series(youth_rates_pct, name="percent")
youth_rates.to_frame()
```

## 7. Registered mobile wallets
The chart labels on p. 4 report registered wallets in millions. Registration does not necessarily imply active use.

```python
wallets_millions = {2019: 15.2, 2020: 19.8, 2021: 25.2, 2022: 30.4, 2023: 39.4, 2024: 50.4, 2025: 60.0}
registered_wallets = pd.Series(wallets_millions, name="million wallets")
registered_wallets.to_frame()
```

## 8. Mobile-wallet transaction value
The report states **EGP 4 trillion** in mobile-wallet transactions during 2025 (p. 4). This is a value flow, distinct from the registered-wallet count.

```python
wallet_transaction_value = pd.Series({2025: 4.0}, name="EGP trillion")
wallet_transaction_value.to_frame()
```

## 9. Build the final CBE context table
Each metric has an explicit unit, segment, definition, and source page. The rate values are stored on a 0–1 scale.

```python
metric_specs = [
    ("financial_inclusion_rate", "Overall financial inclusion rate", "All citizens aged 15+", overall_rates_pct, "proportion", 2,
     "Active transactional accounts at banks, Egypt Post, mobile wallets, or prepaid cards; citizens aged 15+."),
    ("financial_inclusion_rate_women", "Women's financial inclusion rate", "Women aged 15+", women_rates_pct, "proportion", 2,
     "Financially included women under the CBE active transactional account definition."),
    ("financial_inclusion_rate_youth", "Youth financial inclusion rate", "Youth aged 15-35", youth_rates_pct, "proportion", 2,
     "Financially included youth aged 15-35 under the CBE definition."),
    ("registered_mobile_wallets", "Registered mobile wallets", "All registered wallets", wallets_millions, "million wallets", 4,
     "Total registered mobile wallets; registration is not equivalent to active use or unique people."),
    ("mobile_wallet_transaction_value", "Mobile-wallet transaction value", "All mobile-wallet transactions", {2025: 4.0}, "EGP trillion", 4,
     "Value of mobile-wallet transactions during 2025, in Egyptian pounds."),
]
rows = []
for code, name, segment, values, unit, page, definition in metric_specs:
    for year, published_value in values.items():
        value = published_value / 100 if unit == "proportion" else published_value
        rows.append({"country": "Egypt", "year": year, "metric_code": code,
                     "metric_name": name, "segment": segment, "value": value,
                     "unit": unit, "source": "Central Bank of Egypt - Financial Inclusion and Payment Systems & Services Indicators (December 2025)", "source_page": page,
                     "definition_note": definition})

cbe_context = pd.DataFrame(rows)
cbe_context = cbe_context.sort_values(["metric_code", "year"]).reset_index(drop=True)
cbe_context.head()
```

## 10. Data-quality validation
Check the observation key, types, value ranges, provenance, and exact published year coverage. No missing observation is replaced with zero.

```python
rate_codes = {"financial_inclusion_rate", "financial_inclusion_rate_women", "financial_inclusion_rate_youth"}
assert cbe_context.shape == (34, 10)
assert pd.api.types.is_integer_dtype(cbe_context["year"])
assert pd.api.types.is_numeric_dtype(cbe_context["value"])
assert not cbe_context.isna().any().any()
assert not cbe_context.duplicated(["country", "year", "metric_code"]).any()
assert set(cbe_context["country"]) == {"Egypt"}
assert set(cbe_context["metric_code"]) == {spec[0] for spec in metric_specs}
assert cbe_context.loc[cbe_context["metric_code"].isin(rate_codes), "value"].between(0, 1).all()
assert (cbe_context.loc[~cbe_context["metric_code"].isin(rate_codes), "value"] >= 0).all()
for code, _, _, published, unit, page, _ in metric_specs:
    subset = cbe_context[cbe_context["metric_code"] == code].set_index("year")
    assert set(subset.index) == set(published)
    assert set(subset["source_page"]) == {page}
    assert all(abs(subset.at[year, "value"] - (raw / 100 if unit == "proportion" else raw)) < 1e-12
               for year, raw in published.items())
print("Validated", cbe_context.shape, "with zero missing values and zero duplicate observations.")
cbe_context.groupby(["metric_code", "unit"]).agg(first_year=("year", "min"), last_year=("year", "max"), observations=("year", "size"))
```

## 11. Export
Write the clean long table to the existing project’s processed-data folder.

```python
output_path.parent.mkdir(parents=True, exist_ok=True)
cbe_context.to_csv(output_path, index=False)
print(f"Saved {output_path}")
```

## 12. Read-back verification
Reload the CSV to confirm its schema, values, and observation key survived export.

```python
saved = pd.read_csv(output_path)
assert saved.shape == cbe_context.shape
assert saved.columns.tolist() == cbe_context.columns.tolist()
assert not saved.isna().any().any()
assert not saved.duplicated(["country", "year", "metric_code"]).any()
pd.testing.assert_frame_equal(saved, cbe_context, check_dtype=False, rtol=1e-12)
print("Read-back passed:", saved.shape)
saved.tail()
```

## 13. Output summary
The CBE context table has 34 observations across five metrics: overall, women’s, and youth inclusion rates; registered mobile wallets; and 2025 mobile-wallet transaction value. Coverage is 2016–2025 for overall and women, 2020–2025 for youth, 2019–2025 for wallets, and 2025 for transaction value. It remains separate from Findex and IMF FAS.
